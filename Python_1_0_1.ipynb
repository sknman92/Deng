There are additional data types which can contain multiple items, **Lists** and **Dictionaries**

## 2) Packages and Libraries

This is how you can obtain your datetime, tables, dataframes, and more. There's over 500,000 python packages and libraries out there in the wild. But what are they? 

**Library**
- Collection of useful codes you can reuse
  - Functions
  - Classes
  - Tools
  - For example NumPy (Number python) is a library that allows you to create arrays for example (Lists but of all numbers where you can execute advanced mathematical operations)

 
**Packages**
- Collection of related python files (modules) grouped together in a folder
  - For example Pandas is a python package, allowing you to use data frames (tabular data)


Libraries and packages can be public or native

Native don't required installation and just need to be imported in your code such as 
- math (mathematical functions)
- random (generate random numbers)
- datetime (datetime data types, functions etc.)
Non-native need to be installed via tools like pip and then imported in your code

How does one install a public package? 
**This is done in your terminal within an environment**

`pip install [package]`

or
`pip install [package1], [package2], [package3]`

Once installed you have to import it to your code as: 

`import [package] as [abbreviation]`

You can then use functions from the package as:

`abbreviation.function`


Let's use datetime for example: 

````python
import datetime as dt


#Create a date variable
date = dt.datetime(2025, 1, 11, 9, 0,0)
print(date)

#The function is called datetime.now hence dt.datetime.now() instead of dt.now()
print(dt.datetime.now())
````

## 3) Loops

There are two main types of loops: 

**For Loops**

Syntax:

```python
for x in y:
    code block referencing x


- Keep doing a task until you have exhausted the list i have provided
- Automate repetitive tasks
- Makes code shorter and avoid repeating yourself
- **Indentation is important!!!!!!**
- Syntax has 3 key features as `for x in y:
  - **for** : tells python you are starting a foor loop
  - **x**: what you call this parameter is up to you, it's what you would call one item in the list **y**
  - **y**: this is the collection of items you want to iterate thought, or loop through
- The code block referencing x needs to be indented, all lines.

For example:

````python
#Creating a list
shopping_list = ["milk", "bread", "eggs"] 

#Looping through list
for item in shopping_list:
    #each time print the item to buy
    print(f'Remember to buy {item}')
````

````python
#Creating a list
numbers = [4,8,15,16, 23, 42]
#Define variable before the for loop!
total =0

#loop through the numbers
for num in numbers:
    #sum up our numbers, += operator is an easier way to write total = total + number here
    total+= num

print(total)
````

````python
#You can also loop through a range of numbers: 
#range(x,y) creates a list of numbers to loop through, the second number is NOT INCLUDED
for number in range(1,6):
    print(number*number)
````

````python
#Loop through a list of dictionaries 
#Creating a list of dictionaries
user_dictionary= [
    {"name": "Jenny", "active": True},
    {"name": "James", "active": False},
    {"name": "Jules", "active": True}
]

#Loop through the list
for user in user_dictionary:
    #If statement to only print active users, if user["active"] is equivalent to if user["active"] = True
    if user['active'] == True:
        print(f"{user['name']} is active")

````

````python
for entry in user_dictionary:
    if entry['active'] == True:
        print(f"{entry['name']} is active")
    
````

**While Loops**

Syntax:

```python
while condition:
    code block 


- Keep doing a task until the condition is FALSE
- Automate repetitive tasks
- Makes code shorter and avoid repeating yourself
- **Indentation is important!!!!!!**
- Syntax has 2 key features as while condition:
  - **while** : tells python you are starting a while loop
  - **condition**: set your condition that will break the loop, as long as it's true the loop will repeat, as soon as this becomes false, the code will stop iterating
- The whole code block needs to be indented or it will be considered out of the loop
- Within a while loop you can add a break or continue statements, break will lead you out of the loop, continue will skip an iteration and keep the while loop iterating through the next run, you can use these powerfully along if statements.

For example:

````python
#Create a count variable
count = 0

#Looping until count is over 5
while count <= 5: 
    count+=1
    if count == 3:
        #Use continue statement, notice how 3 isn't printed? 
        continue
    print(count)
    
````

````python
#Password example
#Creating a password variable
password = ''
#Creating a count
count = 1

#Until the password is right, keep trying
while password != "python123":
    #Input() asks for a user input, run the cell to see and type whatever you want or the password
    password = input("Enter the password: ")
    #if you fail more than 3 times break the while loop
    if count > 2:
        #Break leads you out of the while loop stopping iterations
        break
        
    count +=1

#Based on the count number we can now if the access was granted or not
if count > 2:
    print("Access denied!")
else:
    print("Access granted")
````

## 4) Functions

A function is a reusable block of code that performs a repeated task. You define the function once and then you can use or call it as many times as you want and even across different python scripts. 

Syntax:

```python
def function_name (parameter 1, parameter 2....):
    code block reference parameter 1 
    code block reference parameter 2 
    return result/output of the function


- Makes code cleaner and easier to read
- Avoids repetition
- Helps with testing and debugging
- Makes your code modular and reusable
- Syntax includes 4 key features
  - **def**: begins the function
  - **function_name**: name of your function, make it easy to understand, you will need to use it whenever you want to run the function
  - **parameters**: optional inputs that your functions uses (you can make them necessary to run the functions or optional with a default value in case there is no value inputed)
  - **return**: optional modified that returns an output of your function
 
For Example: 

````python
#Addition Function

def add(a,b):
    return a+b
print(add(4,5))
````

````python
#Currency conversion

#set function with default rate value)
def euros_to_pound(eur, rate=0.87):
    return eur*rate

#Print with default rate
print(euros_to_pound(100))

#Print with set rate
print(euros_to_pound(100,0.75))
````

````python
#Categorisation

def categorise_spend(amount, low_limit=50, high_limit = 200):
    if amount < low_limit:
        return 'Low Spending'
    elif low_limit <= amount < high_limit:
        return 'Medium Spending'
    else:
        return 'High Spending'

print(categorise_spend(150, low_limit = 200, high_limit = 400))
````

````python
#Circle Area

from math import pi

def circle_area(radius: float = 5):
    #assert isinstance(radius, (float)), "Please enter a valid number"
    area = radius**2 * pi
    return area

circle_area(3.2)
````

````python
#Celcius to Farenheits 

def temp_converter(celsius: float = 32):
    f = (celsius * 1.8) + 32
    return f

temp_converter(22)
````
